# Quickstart: Generate Reports

This notebook demonstrates generating HTML/PDF reports from backtest results.

```python
from backtest.engine import BacktestEngine
from strategies.mean_reversion import MeanReversionStrategy
from data.loader import DataLoader
from utils.reporting import generate_html_report, generate_pdf_report, generate_report

loader = DataLoader()
data = loader.load_data(['BTC'], '2024-01-01', '2024-02-01', freq='1h')
data = loader.clean_data(data)
engine = BacktestEngine(initial_capital=100000.0)
strat = MeanReversionStrategy(asset='BTC')
result = engine.run(strat, data)
print(generate_report(result))
```

```python
html = generate_html_report(result)
with open('../reports/report.html', 'w') as f:
    f.write(html)
print('Saved reports/report.html')
```

```python
try:
    generate_pdf_report(result, '../reports/report.pdf')
    print('Saved reports/report.pdf')
except Exception as e:
    print(e)
```
