# VWAP/TWAP Execution Demo
Demonstrates order execution simulation with VWAP and TWAP fills.

```python
from __future__ import annotations
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
import pandas as pd
from data.loader import DataLoader
from strategies.mean_reversion import MeanReversionStrategy
from backtest.engine import BacktestEngine
from backtest.execution import ExecutionSimulator
```

```python
loader = DataLoader()
data = loader.load_data(['BTC','ETH'], '2024-01-01','2024-02-01', freq='1h')
data = loader.clean_data(data)
data.head()
```

```python
sim = ExecutionSimulator(order_type='vwap', participation_rate=0.1, slippage_bps=5)
engine = BacktestEngine(initial_capital=100_000)
strat = MeanReversionStrategy(asset='BTC')
result = engine.run(strat, data, execution_simulator=sim)
print(result['total_return'], result['sharpe_ratio'])
```
